```python
import numpy as np
from scipy.optimize import brentq

def t_domain_edge(gamma):
    """Upper t bound for real solutions. None if gamma <= 2 (no restriction)."""
    if gamma <= 2:
        return None
    u_minus = 1 - np.sqrt(1 - 2/gamma)
    return 2 * u_minus**2

def residual(t, B, alpha, beta, gamma):
    num = 1/gamma - np.sqrt(t/2) + t/4
    den = 1/gamma + np.sqrt(t/2) + t/4
    if num <= 0:
        return np.nan          # outside domain of real solutions
    return t**(-B) - alpha*beta*(num/den)**(1/gamma)

def solve_t(B, alpha, beta, gamma, n_scan=4000, t_hi_fallback=1e3):
    hi = t_domain_edge(gamma)
    hi = hi*0.999999 if hi is not None else t_hi_fallback  # raise if t could be larger
    grid = np.geomspace(1e-4, hi, n_scan)
    vals = np.array([residual(x, B, alpha, beta, gamma) for x in grid])
    finite = np.isfinite(vals)
    roots = []
    for i in range(len(grid) - 1):
        if finite[i] and finite[i+1] and vals[i]*vals[i+1] < 0:
            roots.append(brentq(residual, grid[i], grid[i+1], args=(B, alpha, beta, gamma)))
    return roots   # inspect all of these before picking one

t_roots = solve_t(B=1.04, alpha=1.21586*10**-6, beta=127451, gamma=6.82256)
print(t_roots)
```
