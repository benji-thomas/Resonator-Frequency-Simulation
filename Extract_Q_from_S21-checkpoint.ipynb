# Quality Factor Extraction from S21 Data
This notebook loads S21 data from a CSV file, converts the magnitude and phase into a complex S21 signal, and fits a resonator model using `resonator_tools` to extract Q values.

```python
# Install the package (uncomment if not already installed)
# !pip install git+https://github.com/sebastianprobst/resonator_tools.git
```

```python
import pandas as pd
import numpy as np
from resonator_tools import circuit
```

## Load and Prepare the Data

```python
# Replace the filename with your actual file path
file_path = 'TWPA_on_S21_4.91-4.93GHz_-25.0dBm_20250605_205559.csv'
df = pd.read_csv(file_path)

# Convert units
freq = df['Frequency (GHz)'].values * 1e9
mag_lin = 10**(df['Magnitude (dB)'].values / 20)
phase_rad = np.deg2rad(df['Phase (deg)'].values)
s21_complex = mag_lin * np.exp(1j * phase_rad)
```

## Fit Resonator Model and Extract Q Values

```python
fit = circuit.notch_port()
fit.add_data(freq, s21_complex)
fit.do_calibration(method='circle')
fit.circlefit()

print(f"Ql = {fit.Ql:.2e}")
print(f"Qc = {fit.Qc:.2e}")
print(f"Qi = {fit.Qi:.2e}")
```
