# Lorentzian peak plots

Two figures:
1. A single Lorentzian centred at `f0`, with the x-axis labelled only at the peak and a dashed guide line down to the axis — sized to drop cleanly into a PowerPoint slide.
2. The same Lorentzian plus two copies shifted by `+delta_f` and `-delta_f`, coloured red and blue, with all three peaks labelled.

Adjust `f0`, `gamma` (linewidth) and `delta_f` below to match your data.

```python
import numpy as np
import matplotlib.pyplot as plt

def lorentzian(f, f0, gamma, amplitude=1.0):
    """Lorentzian lineshape: amplitude * gamma^2 / ((f - f0)^2 + gamma^2)."""
    return amplitude * gamma**2 / ((f - f0)**2 + gamma**2)

# --- Parameters (edit these) ---
f0 = 5.0       # peak / centre frequency
gamma = 0.5    # half-width at half-maximum (HWHM)
```

```python
# --- Figure 1: single Lorentzian, PowerPoint-friendly ---
f = np.linspace(f0 - 6 * gamma, f0 + 6 * gamma, 1000)
y = lorentzian(f, f0, gamma)
peak = lorentzian(f0, f0, gamma)

fig, ax = plt.subplots(figsize=(10, 5))  # 2:1 aspect, fits a slide nicely

ax.plot(f, y, color="black", linewidth=2)

# dashed line from the peak down to the x-axis
ax.plot([f0, f0], [0, peak], linestyle="--", color="black", linewidth=1)

# x-axis: only one label, at f0
ax.set_xticks([f0])
ax.set_xticklabels([r"$f_0$"], fontsize=14)

# no y-axis clutter, no title
ax.set_yticks([])
ax.set_ylim(0, peak * 1.15)
ax.set_xlim(f.min(), f.max())

for spine in ["top", "right", "left"]:
    ax.spines[spine].set_visible(False)

plt.tight_layout()
plt.show()

# fig.savefig("lorentzian_single.png", dpi=300, transparent=True)  # uncomment to export
```

```python
# --- Figure 2: original + two shifted copies (+/- delta_f) ---
delta_f = 1.5  # shift, in the same units as f0/gamma

f2 = np.linspace(f0 - delta_f - 6 * gamma, f0 + delta_f + 6 * gamma, 1000)

y_center = lorentzian(f2, f0, gamma)
y_minus = lorentzian(f2, f0 - delta_f, gamma)
y_plus = lorentzian(f2, f0 + delta_f, gamma)

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(f2, y_center, color="black", linewidth=2, label=r"$f_0$")
ax.plot(f2, y_minus, color="blue", linewidth=2, label=r"$f_0-\Delta f$")
ax.plot(f2, y_plus, color="red", linewidth=2, label=r"$f_0+\Delta f$")

# dashed guide lines, colour-matched to each curve
ax.plot([f0, f0], [0, peak], linestyle="--", color="black", linewidth=1)
ax.plot([f0 - delta_f, f0 - delta_f], [0, peak], linestyle="--", color="blue", linewidth=1)
ax.plot([f0 + delta_f, f0 + delta_f], [0, peak], linestyle="--", color="red", linewidth=1)

# x-axis: three labelled peak positions
ax.set_xticks([f0 - delta_f, f0, f0 + delta_f])
ax.set_xticklabels([r"$f_0-\Delta f$", r"$f_0$", r"$f_0+\Delta f$"], fontsize=14)

ax.set_yticks([])
ax.set_ylim(0, peak * 1.15)
ax.set_xlim(f2.min(), f2.max())

for spine in ["top", "right", "left"]:
    ax.spines[spine].set_visible(False)

plt.tight_layout()
plt.show()

# fig.savefig("lorentzian_triplet.png", dpi=300, transparent=True)  # uncomment to export
```
